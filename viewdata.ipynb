```python
import os
import scanpy as sc

DATA_DIR = r"D:\vscode all\fuxian\scCausalVI-reproducibility-main\data"
FILE = "simulated_balanced_data.h5ad"

adata = sc.read_h5ad(os.path.join(DATA_DIR, FILE), backed="r")   # 只读方式打开，不整体载入内存

print(f"形状 (细胞数, 基因数): {adata.shape}")

print(f"\nX 前 5 行 x 前 5 列:\n{adata.X[:5, :5]}")
print(f"\nobs 前 5 行:\n{adata.obs.head().to_string()}")
print(f"\nvar 前 5 行:\n{adata.var.head().to_string()}")

print(f"\n细胞名前 5 个: {list(adata.obs_names[:5])}")
print(f"基因名前 5 个: {list(adata.var_names[:5])}")

print(f"\nobsm（降维/嵌入）: {list(adata.obsm.keys())}")
print(f"layers（额外矩阵）: {list(adata.layers.keys())}")
print(f"uns（非结构化信息）: {list(adata.uns.keys())}")

print(f"\ncell_type 分布:\n{adata.obs['cell_type'].value_counts().to_string()}")

adata.file.close()   # backed 模式用完要关闭文件句柄
```

```python
import h5py

def dump(name, obj):
    if isinstance(obj, h5py.Dataset):
        print(f"{name}  shape={obj.shape}  dtype={obj.dtype}")
    else:
        print(f"{name}/  (group)")

with h5py.File("simulated_balanced_data.h5ad", "r") as f:
    f.visititems(dump)
```
